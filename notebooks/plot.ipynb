## Line plots (Fig2)

```python
import numpy as np
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'
export_dir = os.getcwd()
import pickle
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path
```

```python
data_name = "ML1M" ### Can be ML1M, Yahoo, Pinterest
recommender_name = "MLP" ## Can be MLP, VAE
task="Top1"   ## Can be "Top1" and "Top10"
```

## Loading records of explainers

```python

with open(Path(Path(os.getcwd(),'scripts'),f'checkpoints/Records_LXR10_{task}_{data_name}_{recommender_name}.pkl'), 'rb') as file:
    Records_LXR10 = pickle.load(file)
MPNR_LXR10=[]
for i in Records_LXR10:
    MPNR_LXR10.append(i['MPNR'])




with open(Path(Path(os.getcwd(),'scripts'),f'checkpoints/Records_LXR_{task}_{data_name}_{recommender_name}.pkl'), 'rb') as file:
    Records_LXR = pickle.load(file)
MPNR_LXR=[]
for i in Records_LXR:
    MPNR_LXR.append(i['MPNR'])




with open(Path(Path(os.getcwd(),'scripts'),f'checkpoints/Records_POP_{task}_{data_name}_{recommender_name}.pkl'), 'rb') as file:
    Records_POP = pickle.load(file)
MPNR_POP=[]
for i in Records_POP:
    MPNR_POP.append(i['MPNR'])




with open(Path(Path(os.getcwd(),'scripts'),f'checkpoints/Records_LF_{task}_{data_name}_{recommender_name}.pkl'), 'rb') as file:
    Records_LF = pickle.load(file)
MPNR_LF=[]
for i in Records_LF:
    MPNR_LF.append(i['MPNR'])
```

## Ploting line plots

```python

# Count the occurrences of each unique value for the last epoch of training
LXR10_frequency = Counter(MPNR_LXR[79])  ## 79 is the MPNR list for the last training epoch (its not a magic number)
LXR_frequency = Counter(MPNR_LXR10[79])  ## 79 is the MPNR list for the last training epoch (its not a magic number)
LF_frequency = Counter(MPNR_LF)
POP_frequency = Counter(MPNR_POP)


# Prepare the x (number of items removed) and y (number of users) axes
x_LXR = sorted(LXR_frequency.keys())
y_LXR = [LXR_frequency[key] for key in x_LXR]

x_LXR10 = sorted(LXR10_frequency.keys())
y_LXR10 = [LXR10_frequency[key] for key in x_LXR10]


x_LF = sorted(LF_frequency.keys())
y_LF = [LF_frequency[key] for key in x_LF]

x_POP = sorted(POP_frequency.keys())
y_POP = [POP_frequency[key] for key in x_POP]

# Compute the cumulative count
y_cumulative_LXR = np.cumsum(y_LXR)
y_cumulative_LXR10 = np.cumsum(y_LXR10)
y_cumulative_LF = np.cumsum(y_LF)
y_cumulative_POP= np.cumsum(y_POP)



# Plot the cumulative count
plt.plot(x_POP, y_cumulative_POP, linestyle='-', color='g', label="POP")
plt.plot(x_LF, y_cumulative_LF, linestyle='-', color='r', label="LF")
plt.plot(x_LXR, y_cumulative_LXR, linestyle='-', color='b', label="LXR")
plt.plot(x_LXR10, y_cumulative_LXR10, linestyle='-', color='k', label="LXR10")

# Customize the plot
plt.xlabel('MPNR')
plt.ylabel('Cumulative Percentage of Users')
plt.title(f'Recommender:{recommender_name}, Dataset:{data_name}')
plt.legend()
plt.xlim(0,25)
plt.savefig(f"{recommender_name}-{data_name}.pdf")
# Show the plot
plt.show()
```
